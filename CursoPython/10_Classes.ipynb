```python
import random
#Creiando a Classe
class peDeLaranja:

    def __init__ (self):
        self._altura = 0
        self._idade = 0
        self._frutos = 0
        self._viva = True

    def adicionar_um_ano(self):
        self._idade += 1
        self._adicionar_um_metro()
        self._adicionar_frutos()
        self._continuar_viva()
        print(f'{self._idade} anos, {self._altura}m, {self._frutos} futas e está viva (?){self._viva}')

    def _adicionar_um_metro(self):
        if self._altura < 10:
            self._altura +=1

    def _adicionar_frutos(self):
        if (self._idade >= 5) & (self._idade <= 10):
            self._frutos += 100
        elif (self._idade >= 11) & (self._idade <= 15):
            self._frutos += 200

    def colher_frutos(self, qtde):
        if qtde <= self._frutos:
            self._frutos -= qtde
        else:
            self._frutos = 0

    def _continuar_viva(self):
        if self._idade >= 100:
            self._viva = False
        elif (self._idade >= 50) & (random.randint(0,49)==0):
            self._viva = False

```

```python
#Instanciando a classe
laranjaLima = peDeLaranja()
```

```python
laranjaLima.adicionar_um_ano()
#Verificando Atributos
laranjaLima._altura, laranjaLima._idade, laranjaLima._frutos, laranjaLima._viva
```

```python
laranjaLima.colher_frutos(300)
#Verificando Atributos
laranjaLima._altura, laranjaLima._idade, laranjaLima._frutos, laranjaLima._viva
```
