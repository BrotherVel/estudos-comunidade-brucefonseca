#### Exemplos: Funções de Iteração

```python
# range() - Sequência de números
print("Range de 0 a 5:")
for numero in range(5):
    print(numero, end=" ")
print()

# range com início e fim
print("\nRange de 2 a 7:")
for numero in range(2, 7):
    print(numero, end=" ")
print()

# range com passo
print("\nRange de 0 a 10 (passo 2):")
for numero in range(0, 10, 2):
    print(numero, end=" ")
print()

# map() - Aplica função
numeros = [1, 2, 3, 4]
dobrados = list(map(lambda x: x * 2, numeros))
print(f"\nOriginal: {numeros}")
print(f"Dobrados: {dobrados}")

# filter() - Filtra elementos
pares = list(filter(lambda x: x % 2 == 0, numeros))
print(f"Pares: {pares}")

# zip() - Combina sequências
lista1 = [1, 2, 3]
lista2 = ["a", "b", "c"]
combinadas = list(zip(lista1, lista2))
print(f"Combinadas: {combinadas}")
```

#### Exemplos: Funções Úteis

```python
# isinstance() - Verifica tipo
valor = 42
print(f"42 é inteiro? {isinstance(valor, int)}")
print(f"42 é string? {isinstance(valor, str)}")

lista = [1, 2, 3]
print(f"[1,2,3] é lista? {isinstance(lista, list)}")

# all() - Todos elementos verdadeiros
lista_true = [True, True, True]
lista_mista = [True, False, True]
print(f"\nTodos True em {lista_true}? {all(lista_true)}")
print(f"Todos True em {lista_mista}? {all(lista_mista)}")

# any() - Algum elemento verdadeiro
print(f"Algum True em {lista_true}? {any(lista_true)}")
print(f"Algum True em {lista_mista}? {any(lista_mista)}")

lista_false = [False, False]
print(f"Algum True em {lista_false}? {any(lista_false)}")
```

#### Escopo de variáveis em funções

```python
numero = 5

def definir_numero(valor):
    numero = valor
    return numero

print(definir_numero(10))
print(numero)

# O numero dentro da função, a priori, não é o mesmo que o numero fora da função. 
# A função cria um escopo local, e a variável numero dentro da função é diferente da variável numero fora da função.
#  Portanto, a alteração feita dentro da função não afeta a variável global numero.
```

```python
numero = 5

def definir_numero(valor):
    global numero # O global indica que a variável numero dentro da função se refere à variável global numero
    numero = valor
    return numero


print(definir_numero(10))
print(numero)

# Desta forma a alteração feita dentro da função afeta a variável global.
```

```python
mensagem1 = "Variável Global"

def minhaFuncao():
    print(mensagem1) # A função consegue acessar a variável global mensagem1

    mensagem2 = "Variável Local"
    print(mensagem2) # A função consegue acessar a variável local mensagem2


try:
    print(mensagem2) # A variável mensagem2 não é acessível fora da função, pois é uma variável local.
except NameError as e:
    print(f"Erro: {e}")
```

##### Função 'items()'

```python
precos_manutencao = {
    "troca de óleo": 50,
    "Filtro de ar": 20,
    "Pastilhas de freio": 80,
    "bateria": 100,
    "alinhamento": 60,
}

contagem = 0

print("Forma rudimentar de acessar os valores do dicionário")
for valor in precos_manutencao:
    print(f'serviço: {valor}')  # Acessa a chave do dicionário
    print(f'preço: {precos_manutencao[valor]}')  # Acessa o valor do dicionário usando a chave

print()
print("Desta forma, cada valor é uma tupla (chave, valor)")
for valor in precos_manutencao.items():
    print(valor)  # desta forma, cada valor é uma tupla (chave, valor)

print()

print("Acessa a chave e o valor diretamente de forma individual")
for direita, esquerda in precos_manutencao.items():
    print(f"Serviço: {direita}")
    print(f"Preço: {esquerda}")  # Acessa a chave e o valor diretamente de forma individual

print()
print("Acessa apenas o valor do dicionário")
for valor in precos_manutencao.values():
    print(f"Preço: {valor}")  # Acessa apenas o valor do dicionário, ao contrário da forma padrão
```

#### Criando funções

```python
#Criando um dicionario com preços de manutenção 

precos_manutencao = { 
    "Troca de Óleo": 50, 
    "Filtro de Ar": 20, 
    "Pastilhas de Freio": 80, 
    "Bateria": 100,
    "Alinhamento": 60
}

# Função para calcular o custo total da manutenção

pedido_servico = ["Pastilhas de Freio", "Alinhamento"]

def calcular_precos(pedidos):

    valor_total = 0

    for servico, precos in precos_manutencao.items():
        if servico in pedidos:
            valor_total += precos

    return valor_total

calcular_precos(pedido_servico)
```

```python
#Criando um dicionario com tempos de manutenção 

tempo_manutencao = { 
    "Troca de Óleo": 30, 
    "Filtro de Ar": 15, 
    "Pastilhas de Freio": 45, 
    "Bateria": 60,
    "Alinhamento": 40
}

# Função para calcular o custo total da manutenção

def calcular_tempo(pedidos):

    tempo_total = 0
    for pedido in pedidos:
        for servico, tempo in tempo_manutencao.items():
            if pedido == servico:
                tempo_total += tempo

    return tempo_total
```

```python
# Função principal para fazer o pedido de manutenção
def validacao(verificar):
    
    validado = False

    if verificar in precos_manutencao :
        validado = True
    
    return validado

# coverter min em horas

def calcularHoras(tempo):
    if tempo > 60:
        
        horas = tempo // 60
        min = tempo % 60

        return f"{horas}h{min}m"
    else:
        return tempo

# Função mestre

def Servico_Manutencao():
    print("Lista de Serviços de Manutenção:")

    contador = 1

    for lista in precos_manutencao:
        print(contador, "-", lista)
        contador += 1

    deseja_contratar_servico = input("Deseja contratar algum serviço? (s/n)")
    print("Deseja contratar algum serviço? (s/n)")
    print(deseja_contratar_servico)

    while True:

        if deseja_contratar_servico == "s":

            pedidos_servicos = []

            while True:

                pedido = input("Digite o nome do serviço")

                pedidos_servicos.append(str(pedido)) 
                
                validado = validacao(pedido)

                if validado == True:

                    print(f"Foi adicionado o serviço \'{pedido}\' ao carrinho")
                    continuar = "s"
                    print("Deseja adicionar mais algum serviço? (s/n)")
                    continuar = input("Deseja adicionar mais algum serviço? (s/n)")             

                    if continuar != "s":
                        print("Então vamos continuar para a próxima etápa !!")
                        break
                    else:
                        print("Sim ?! então adicione o proximo serviço desejado :D")
                    
                else:
                    print("Não temos esse serviço :( insira novamente")
                    pedidos_servicos.remove(pedido)

            print()
            print("Consulte a tabela com o valor e o tempo de manutenção totais:") 
            print()
            print(f"{'Índice':<10} | {'Item':<20} | {'Preço':<13} | {'Tempo ':<10}")
            print("-" * 70)

            indice = 1

            for item in pedidos_servicos:

                print(f"{indice:<10} | {item:<20} | R$ {precos_manutencao[item]:<10} | {tempo_manutencao[item]} min")
                indice += 1

            print("-" * 70)

            time = calcular_tempo(pedidos_servicos) 
            price_total = calcular_precos(pedidos_servicos)
            print(f"{' ':<10} | {'Total':<20} | R$ {price_total:<10} | {time}min")
            
            break

        elif deseja_contratar_servico == "n":
            print("Obrigado por consultar nossos serviços!!!")
            break
        else:
            print("insira um valor válido")
            deseja_contratar_servico = input("Deseja contratar algum serviço? (s/n)")

Servico_Manutencao()
```
