#### Triple Quotes

```python
#""" * """ 

texto = """Lorem ipsum dolor sit amet, 
consectetur adipiscing elit. 
Donec a diam lectus. Sed sit amet ipsum mauris."""

print(texto) #O print de um texto guardado em aspas duplas triplas consegue interpretar as quebras de linha e os espaços, ou seja, a formatação do texto é mantida.
```

#### Escape Characters

```python
# \t = tab (tabulação)

print("Python é uma linguagem de programação \t muito fácil de aprender e usar.") 
#O \t é um caractere de escape que representa uma tabulação, ou seja, ele adiciona um espaço em branco equivalente a um tab.
```

```python
# \n = nova linha (quebra de linha)

print("Python é uma linguagem de programação \nMuito fácil de aprender e usar.")
#O \n é um caractere de escape que representa uma nova linha, ou seja, ele quebra a linha e o texto seguinte é impresso na linha abaixo.
```

```python
# \\ = barra invertida (caractere de escape)

print("O caminho para o arquivo é d:\\Progran_Working\\Python_-_CursoBruce_-_Python_Fundamentos\\CursoPython\\variaveis.py")
#O \\ é um caractere de escape que representa uma barra invertida, ou seja, ele permite que a barra seja interpretada como um caractere literal em vez de um caractere de escape.
```

```python
# \" = " (aspas duplas)

print("Eu tenho 5' 10\" de altura.")
#O \" é um caractere de escape que representa uma aspa dupla, ou seja, ele permite que a aspa seja interpretada como um caractere literal em vez de um caractere de escape. 
```

```python
# \' = ' (aspas simples)

print('Eu tenho 5\' 10" de altura.')
#Da mesma forma que o \", o \' é um caractere de escape que representa uma aspa simples.
```

# Lista de escape em Python

\n → nova linha<br>
\t → tabulação<br>
\r → retorno de carro<br>
\b → retrocesso<br>
\f → avanço de página<br>
\v → tabulação vertical<br>
\a → campainha (alerta)<br>
\\ → barra invertida \<br>
\' → apóstrofo simples<br>
\" → aspas duplas<br>
\0 → caractere nulo<br>
<br>
Escape numérico / Unicode<br>
<br>
\ooo → valor octal (ex: \101 = A)<br>
\xhh → valor hexadecimal (ex: \x41 = A)<br>
\uXXXX → Unicode de 4 dígitos (ex: \u00E7 = ç)<br>
\UXXXXXXXX → Unicode de 8 dígitos<br>
\N{name} → caractere Unicode por nome (ex: \N{HEAVY BLACK HEART})
