#### Formatação de Strings

```python
marca = "iphone 15"
lancamento = 2023
valor = 1299
taxaDolar = 4.95555555

#Forma Não otimizada:
print('O ' + marca + ' foi lançado em ' + str(lancamento))
print()

#Forma otimizada
print('O %s foi lançado em %d custano %d. Mas a taxa de consão de real para dolar é de R$%.2f para 1 dolar'%(marca, lancamento, valor, taxaDolar)) 
#As marcações são deixadas no texto para representarem as variáveis que podem ser inseridas posteriormente.

#Maracações para cada tipo de dados:
#string ->  %s
#Int ->     %d
#Float ->   %.(número de casas decimais)f 
```

#### Format

```python
marca = "iphone 15"
lancamento = 2023
valor = 1299
taxaDolar = 4.95555555

#Opção utilizando o format:
print('O {0} foi lançado em {1} custano {2}. Mas a taxa de consão de real para dolar é de R${3:.2f} para 1 dolar'.format(marca, lancamento, valor, taxaDolar))
# Funciona atravez de um índice que conta a partir do 0 e pode ser atribuido respectivamente pela ordem estabelecida dentro da função.
# É possivel controlar as casas decimais do tipo float utilizando o ":" seguida da mesma abordagem anterior.
```

```python
#Outra alternativa com o format:
print(f'O {marca} foi lançado em {lancamento} custano {valor}. Mas a taxa de consão de real para dolar é de R${taxaDolar:.2f} para 1 dolar')
# Nesse método, ao invés de utilizar um ídice, utiliza-se a inserção da própria variável.
```
