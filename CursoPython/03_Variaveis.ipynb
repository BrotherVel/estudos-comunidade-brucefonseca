#### Operadores matemáticos:

```python
# "+" Adicção
# "-" Subtração
# "*" Multiplicação
# "/" Divisão Com Decimais (Float)
# "//" Divisão Número Inteiro (Arrendonda para baixo) 
# "%" Resto (Retorna o resto da divisão)
# "**" Potência (Retorna o número elevado ao outro)
```

#### Subtração:

```python
x, y = 4, 3

resultado = x - y

resultado
```

#### Adição:

```python
x, y = 4, 3

resultado = x + y

resultado
```

#### Multiplicação:

```python
x, y = 4, 3

resultado = x * y

resultado
```

#### Divisão Com Decimais (retorna float):

```python
x, y = 5, 3

resultado = x / y

resultado
```

#### Divisão Número Inteiro:

```python
x, y = 5, 3

resultado = x // y

resultado
```

```python
x, y = 5, 3

resultado = x % y

resultado
```

#### Potência (Retorna o número elevado ao outro):

```python
x, y = 4, 3

resultado = x ** y

resultado
```

#### Operadores de atribuição:

```python
# "=" Recebe:

x = 5
x = x + 5

print(x)
```

```python
# "+=" Adiciona e recebe (equivalente ao exemplo anterior)

x += 5  # Equivalente a x = x + 5
#Significa que ele irá receber o valor atribuido de x (que é "10" no exemplo) mais o valor que está recebendo ("5"), 
#praticamente somando o novo valor recebido ao que a váriavel possui ao invés de simplismente substituir.
x
```

```python
# "-=" Subtrai e recebe (equivalente ao exemplo anterior para subtração)

x -= 5 # Equivalente a x = x - 5
x
```

```python
# "*=" Multiplica e recebe (equivalente ao exemplo anterior para multiplicação)

x *= 5 # Equivalente a x = x * 5
#multiplica pelo valor ja atribuido na variável que está recebendo
x
```

```python
# "/=" Divide recebe (equivalente ao exemplo anterior para Divisão)

x /= 5 # Equivalente a x = x / 5
x
```
