#### IF & Else

```python
#if & else

Nota = 7.5

if Nota >= 9:
    print("Parabéns, você foi aprovado com louvor!")
else:
    print("Infelizmente, você não foi aprovado.")

Nota += 2

if Nota >= 9:
    print("Parabéns, você foi aprovado com louvor!")
else:
    print("Infelizmente, você não foi aprovado.")
```

```python
# if sem else

Nota = 7.5

if Nota >= 9:
    print("Parabéns, você foi aprovado com louvor!")
```

```python
#elif

Nota = 5

if Nota >=9:
    print("você recebeu nota A")
elif Nota >= 7:
    print("você recebeu nota B")
elif Nota >= 5:
    print("você recebeu nota C")
else:
    print("Infelizmente, você não foi aprovado.")
```

```python
# in

#  em uma rifa com 10 numeros entre 5 e 20, escolha um numero para ser sorteado

numero = 20

rifa = [19, 5, 17, 11, 12, 10, 15, 20, 8, 12]

if numero in rifa:
    print(f"Parabéns, o número {numero} foi sorteado!")
else:
    print(f"Infelizmente, o número {numero} não foi sorteado.")
```

```python
# or = |
#  em uma rifa com 10 numeros entre 5 e 20, escolha um numero para ser sorteado

numero = 180

rifa = [19, 5, 17, 11, 12, 10, 15, 20, 8, 12]

if numero < 5 or numero > 20:
    print("Número inválido, escolha um número entre 5 e 20.")
elif numero in rifa:
    print(f"Parabéns, o número {numero} foi sorteado!")
else:
    print(f"Infelizmente, o número {numero} não foi sorteado.") 
```

```python
# and = &

numero = 21

rifa = [19, 5, 17, 11, 12, 10, 15, 20, 8, 12]

if numero >= 5 and numero <= 20:
    if numero in rifa:
        print(f"Parabéns, o número {numero} foi sorteado!")
    else:
        print(f"Infelizmente, o número {numero} não foi sorteado.")
else:
    print("Número inválido, escolha um número entre 5 e 20.")
```

```python
# Outra para o 'and' com o &

numero = 4

rifa = [19, 5, 17, 11, 12, 10, 15, 20, 8, 12]

if (numero >= 5) & (numero <= 20):
    if numero in rifa:
        print(f"Parabéns, o número {numero} foi sorteado!")
    else:
        print(f"Infelizmente, o número {numero} não foi sorteado.")
else:
    print("Número inválido, escolha um número entre 5 e 20.")
```

#### While loop

```python
# Contagem regressiva

contador = 5

while contador > 0:
    print(contador)
    contador -= 1
```

```python
# Soma até um valor

soma = 0
limite = 10

while soma < limite:
    print(soma)
    numero = int(input("Digite um número: "))
    soma += numero

print("A soma dos números digitados é", soma, "no total. Sendo assim, você atingiu o limite de", limite, "para a soma.")
```

```python
# advinhe um numero

numero = 78
tentativa = 0

while True:
    palpite = int(input("Digite um numero entre 1 e 100: "))
    tentativa += 1  
    if palpite == numero:
        print("Parabens! Você acertou na tentativa", tentativa, "o numero correto:", numero)
        break
    elif palpite < numero:
        print("tente um numero menor")
    else:
        print("tente um numero maior")
```

#### Break

```python
cestaDeFrutas = ["maça", "banana", "laranja", "uva", "abacaxi"]

while True:

    selecao = input("Digite o nome de uma fruta: ")
    contagem = 1

    for fruta in cestaDeFrutas:
        
        if fruta == selecao:
            print(f"A {fruta} foi encontrada na posição {contagem} da lista.")
            break
        else: 
            contagem += 1
        
        if contagem > len(cestaDeFrutas):
            print(f"A {fruta} não foi encontrada na lista.")
            break
    
    if fruta == selecao:
        break
    else:
        print("Insira uma fruta válida.")

```

```python
cestaDeFrutas = ["maça", "banana", "laranja", "uva", "abacaxi"]

while True:

    print("Lista de frutas disponíveis:") 
    for fruta in cestaDeFrutas:
        print("->", fruta)
    print("Digite o nome de uma fruta: ") #só para aparecer a mensagem no terminal

    selecao = input("Digite o nome de uma fruta: ")
    contagem = 1
    opção = "n"

    print("Você está procurando por:", selecao)
    print("Buscando...")


    for fruta in cestaDeFrutas:
                
        if fruta == selecao:
            print(f"A {fruta} foi encontrada na posição {contagem} da lista.")
            opcao = input("deseja continuar? (s/n)")

            if opcao == "s":
                break
            else:
                print("Encerrando a busca.")
                break

        else: 
            contagem += 1
        
        if contagem > len(cestaDeFrutas):
            print(f"A fruta {selecao} não foi encontrada na lista.")
            break
    
    if opcao == "s":
        continue
    elif fruta == selecao:
        break
    else:
        print("Insira uma fruta válida.")
```

#### Continue

```python
cestaDeFrutas = ["maça", "banana", "laranja", "uva", "abacaxi"]

for fruta in cestaDeFrutas:
    if fruta == "laranja":
        print("Encontramos a laranja! Vamos pular para a próxima interação.")
        continue
    
    print(fruta)
```

#### Try Except

```python

limite = 10
acumulado = 0

while acumulado < limite:

    try:
        print("Acumulado:", acumulado)
        numero = int(input("Digite um número: "))
        print("Número:", numero)
        acumulado += numero

    except ValueError:
        print("Por favor, digite um número válido.")
```
