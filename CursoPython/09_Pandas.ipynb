```python
# Mudar o tipo de dados na importação 

data_type = { # dicionario que será utilizado para alterar por coluna o tipo de dados 
    'CustomerNumber': str,
    'SalesOrder': str
} 
# lembrando das boas praticas de estruturação dos dados que é transformar as keys em strings 
# # Isto é para que as colunas de id se manterem estáticas e não serem alteradas como valores nas análises

df = pd.read_csv(caminho_arquivo, dtype= data_type) 

# dtype é para definir o tipo das colunas que serão trazidas
# Ao inserir um dicionario neste campo com as especificações montadas acima, ela transforma o tipo dos campos específicos definidos. 

df.dtypes
```

```python
# O pandas tbm tem suporte para o tipo data, mas a sua definição é feita de forma diferente

# Utiliza-se o 'parce_dates' para definir as colunas de datas

df = pd.read_csv(caminho_arquivo, dtype= data_type, parse_dates=['SalesDate']) # poderia ser uma variavel com uma lista das colunas ou uma lista direta desta forma com as colunas a serem tratadas como datas

df.dtypes
```

#### Métodos Aplicadas ao Dataframe

```python
sorocaba_df.tail() # traz os últimos itens da tabela (padrão 5)

# head e tail sao metodos que podem ser utilizados para uma visualização rápida do dataframe, mas não são métodos de filtragem, apenas de visualização
```

```python
soma_vendas_sorocaba = sorocaba_df['Revenue'].sum() # soma todos os valores da coluna Revenue do dataframe sorocaba_df

print(soma_vendas_sorocaba)
```

```python
sorocaba_df['Revenue'].sort_values(ascending=False).head() # Ordena os valores da coluna Revenue do dataframe sorocaba_df em ordem decrescente
```

```python
sorocaba_df.sort_values(by='Revenue', ascending=False).head()
# Ordena pela coluna Revenue do dataframe sorocaba_df em ordem decrescente, mas traz todas as colunas do dataframe
```

```python
maior_venda_sorocaba = sorocaba_df['Revenue'].max() # retorna o maior valor da coluna Revenue do dataframe sorocaba_df
print(maior_venda_sorocaba)
```

```python
menor_venda_sorocaba = sorocaba_df['Revenue'].min() # retorna o menor valor da coluna Revenue do dataframe sorocaba_df
print(menor_venda_sorocaba)
```

```python
media_vendas_sorocaba = sorocaba_df['Revenue'].mean() # retorna a média dos valores da coluna Revenue do dataframe sorocaba_df
print(f"{media_vendas_sorocaba:.2f}")
```

```python
mediana_vendas_sorocaba = sorocaba_df['Revenue'].median() # retorna a mediana dos valores da coluna Revenue do dataframe sorocaba_df

# a diferencia entre media e mediana é que a media é a soma de todos os valores dividido pela quantidade de valores, 
# enquanto a mediana é o valor que está no meio da lista de valores ordenados.

print(mediana_vendas_sorocaba)
```

```python
quantidade_vendas_sorocaba = sorocaba_df['Revenue'].count() # retorna a quantidade de vendas (linhas) da coluna Revenue do dataframe sorocaba_df
print(quantidade_vendas_sorocaba)
```

```python
quantidade_valoresUnicos_vendas_sorocaba = sorocaba_df['Revenue'].nunique()

print(quantidade_valoresUnicos_vendas_sorocaba)
```

```python
valoresUnicos_vendas_sorocaba = sorocaba_df['Revenue'].unique() # retorna os valores únicos da coluna Revenue do dataframe sorocaba_df

print(valoresUnicos_vendas_sorocaba)
```

```python
sorocaba_df['Revenue'].std() # retorna o desvio padrão dos valores da coluna Revenue do dataframe sorocaba_df

# o desvio padrão é uma medida de dispersão que indica o quanto os valores de um conjunto de dados se afastam da média. 
# Quanto maior o desvio padrão, maior a variabilidade dos dados em relação à média.
```

#### Group By

```python
sorocaba_df.groupby('Category')['Revenue'].sum().sort_values(ascending=False)
# groupby é para agrupar por uma coluna, e o ['Revenue'] é para definir a coluna que será somada, 
# e o .sum() é uma função para somar os valores da coluna definida

```

```python
sorocaba_df_group = sorocaba_df.groupby(['BusinessLine', 'Category'])['Revenue'].sum().sort_values(ascending=False) 
# groupby com lista de colunas, para agrupar por mais de uma coluna, e o ['Revenue'] é para definir a coluna que será somada.
sorocaba_df_group
```

```python
sorocaba_df_group.reset_index() # reseta o index do dataframe, transformando os valores do index em colunas do dataframe
```

#### Tabelas Dinâmicas

```python
df.pivot_table(
    index='State', # define a coluna que será utilizada como index (linhas) da tabela dinâmica
    columns='Category', # define a coluna que será utilizada como colunas da tabela dinâmica
    values='Revenue', # define a coluna que será utilizada como valores da tabela dinâmica
    aggfunc='sum', # define a função de agregação que será utilizada para os valores da tabela dinâmica
    margins=True, # define se será exibido o total geral da tabela dinâmica
    margins_name='Total' # define o nome da coluna do total geral da tabela dinâmica
)
```
